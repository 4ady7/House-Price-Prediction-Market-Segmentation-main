## Download the Data

````python
from pathlib import Path
import pandas as pd
import tarfile
import urllib.request

def load_housing_data():
    tarball_path = Path("datasets/housing/housing.tgz")
    if not tarball_path.is_file():
        Path("datasets").mkdir(parents=True, exist_ok=True)
        url = "https://github.com/ageron/data/raw/main/housing.tgz"
        urllib.request.urlretrieve(url, tarball_path)
        with tarfile.open(tarball_path) as housing_tarball:
            housing_tarball.extractall(path="datasets")
    return pd.read_csv(Path("datasets/housing/housing.csv"))

housing = load_housing_data()
````

````python
df1.plot(kind="scatter", x="total_bedrooms", y="median_house_value", grid=True,
             s=housing["total_bedrooms"] / 100, label="total_bedrooms", #here s is short for size and also here the size is being divided by a hundred to keep sizes managable
             c="median_house_value", cmap="jet", colorbar=True, #here c is short for color and each house is colored based on its price
             legend=True, sharex=False, figsize=(10, 7))
plt.show()
````

##### In the previous image, we can see clearly that the frequency of high median house value is less than the frequency of low median house value in reagards to the amount of total bedrooms. According the range of houses that has total bedrooms from 1 to 2000 or more, there are more houses that has cheaper value than there are expensive ones. So houses that has total bedrooms from 1 to 2000 or more are more likely to be cheaper and has less median house value than to have a high median house value
